```python
import os
import pandas as pd

t2012 = pd.read_csv("Tabela5-sem_emprego_2012.csv", sep=";", encoding="utf-8-sig")
t2026 = pd.read_csv("Tabela5-sem_emprego_2026.csv", sep=";", encoding="utf-8-sig")

t2012.columns = t2012.columns.str.strip()
t2026.columns = t2026.columns.str.strip()

t5_comp = t2012.merge(t2026, on=["Sigla", "Código", "Estado"], how="inner")

try:
    df_ibge_raw = pd.read_csv("Tabela4.csv", sep=";", encoding="latin1")
    if len(df_ibge_raw.columns) == 1:
        df_ibge_raw = pd.read_csv("Tabela4.csv", sep=",", encoding="latin1")
except Exception:
    df_ibge_raw = pd.read_csv("Tabela4.csv", sep=",", encoding="latin1")

df_ibge_raw.columns = df_ibge_raw.columns.str.strip().str.replace('\ufeff', '')

if len(df_ibge_raw.columns) >= 3:
    novos_nomes = {
        df_ibge_raw.columns[0]: "Sigla",
        df_ibge_raw.columns[1]: "Código",
        df_ibge_raw.columns[2]: "Estado"
    }
    df_ibge_raw = df_ibge_raw.rename(columns=novos_nomes)

t5_comp["Código"] = t5_comp["Código"].astype(str).str.strip()
df_ibge_raw["Código"] = df_ibge_raw["Código"].astype(str).str.strip()

cruzado = t5_comp.merge(df_ibge_raw, on=["Sigla", "Código", "Estado"], how="inner")

caminho_saida = r"H:\dados30\python-tecTI\dados"
os.makedirs(caminho_saida, exist_ok=True)

cruzado.to_csv(os.path.join(caminho_saida, "geral.csv"), sep=";", index=False, encoding="utf-8-sig")
```
